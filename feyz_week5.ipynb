```python
# Soru 1: Rectangle (Dikdörtgen) Sınıfı
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)

# Örnek kullanım
rect = Rectangle(5, 7)
print("Alan:", rect.area())
print("Çevre:", rect.perimeter())
```

```python
# Soru 2: School (Okul) Sınıfı

class School:
    def __init__(self, name, foundation_year):
        self.name = name
        self.foundation_year = foundation_year
        self.students = []
        self.teachers = {}

    def add_new_student(self, student_name, student_class):
        self.students.append((student_name, student_class))

    def add_new_teacher(self, teacher_name, branch):
        self.teachers[teacher_name] = branch

    def view_student_list(self):
        print("Öğrenci Listesi:")
        for name, cls in self.students:
            print(f"{name} - Sınıf: {cls}")

    def view_teacher_list(self):
        print("Öğretmen Listesi:")
        for name, branch in self.teachers.items():
            print(f"{name} - Branş: {branch}")

# Örnek kullanım
school = School("Anadolu Lisesi", 1990)
school.add_new_student("Ahmet", "10A")
school.add_new_teacher("Ayşe Öğretmen", "Matematik")
school.view_student_list()
school.view_teacher_list()
```

```python
# Soru 3: Shape - Rectangle - Square (Şekil - Dikdörtgen - Kare)

class Shape:
    def __init__(self, width, height):
        self.width = width
        self.height = height

class Rectangle(Shape):
    def calculate_area(self):
        return self.width * self.height

class Square(Shape):
    def calculate_area(self):
        return self.width * self.width  # width ve height eşit

# Örnek kullanım
rect = Rectangle(4, 5)
square = Square(3, 3)

print("Dikdörtgen Alanı:", rect.calculate_area())
print("Kare Alanı:", square.calculate_area())
```

```python
# Soru 4: Vehicle - OffRoadVehicle - SportsCar (Araçlar)

class Vehicle:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

class OffRoadVehicle(Vehicle):
    def __init__(self, make, model, year, four_wheel_drive):
        super().__init__(make, model, year)
        self.four_wheel_drive = four_wheel_drive

class SportsCar(Vehicle):
    def __init__(self, make, model, year, max_speed):
        super().__init__(make, model, year)
        self.max_speed = max_speed

# Örnek kullanım
suv = OffRoadVehicle("Toyota", "Land Cruiser", 2020, True)
sports = SportsCar("Ferrari", "488", 2022, 330)

print(f"SUV: {suv.make}, {suv.model}, {suv.year}, 4x4: {suv.four_wheel_drive}")
print(f"Spor Araba: {sports.make}, {sports.model}, {sports.year}, Maksimum Hız: {sports.max_speed} km/s")
```

```python
# Soru 5: Customer ve Account (Müşteri ve Hesap)

class Customer:
    def __init__(self, name, surname, tc_identification, phone):
        self.name = name
        self.surname = surname
        self.tc_identification = tc_identification
        self.phone = phone

    def display_information(self):
        print(f"Ad: {self.name}, Soyad: {self.surname}, TC: {self.tc_identification}, Tel: {self.phone}")

class Account:
    def __init__(self, customer, account_number, balance=0):
        self.customer = customer
        self.account_number = account_number
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"{amount} TL yatırıldı.")

    def money_check(self, amount):
        if self.balance >= amount:
            self.balance -= amount
            print(f"{amount} TL çekildi.")
        else:
            print("Yetersiz bakiye!")

    def display_balance(self):
        print(f"Bakiye: {self.balance} TL")

# Örnek kullanım
cust = Customer("Ali", "Yılmaz", "12345678901", "0555 555 55 55")
acc = Account(cust, "TR1234567890")

cust.display_information()
acc.deposit(500)
acc.money_check(200)
acc.display_balance()
acc.money_check(400)  # Yetersiz bakiye olacak
```
